### 🟩 Tuần 1: Mô phỏng Ma Trận LED 5x5 micro:bit bằng Python

```python
import numpy as np
import matplotlib.pyplot as plt

# Mô phỏng trái tim LED (Heart Icon) trên ma trận 5x5
heart_matrix = np.array([
    [0, 9, 0, 9, 0],
    [9, 9, 9, 9, 9],
    [9, 9, 9, 9, 9],
    [0, 9, 9, 9, 0],
    [0, 0, 9, 0, 0]
])

plt.figure(figsize=(4, 4))
plt.imshow(heart_matrix, cmap='Reds', vmin=0, vmax=9)
plt.title('BBC micro:bit 5x5 LED Matrix Simulation')
plt.xticks(range(5))
plt.yticks(range(5))
plt.grid(True, color='gray', linestyle='--', linewidth=0.5)
plt.show()
```

### 📈 Tuần 6: Phân Tích File Datalogger CSV từ micro:bit v2

```python
import pandas as pd

# Giả lập chuỗi dữ liệu Datalogger CSV thu được từ micro:bit v2
csv_data = """Time_s,Temp_C,Light_Lvl,Soil_Moisture
0.0,26.5,120,65
1.0,26.6,122,64
2.0,26.8,118,62
3.0,27.1,115,55
4.0,27.5,110,48
5.0,28.0,105,25
"""

import io
df = pd.read_csv(io.StringIO(csv_data))

plt.figure(figsize=(8, 4))
plt.plot(df['Time_s'], df['Temp_C'], label='Temp (°C)', color='red', marker='o')
plt.plot(df['Time_s'], df['Soil_Moisture'], label='Soil Moisture (%)', color='blue', marker='s')
plt.xlabel('Time (seconds)')
plt.ylabel('Value')
plt.title('micro:bit v2 Datalogger CSV Telemetry')
plt.legend()
plt.grid(True)
plt.show()
```
