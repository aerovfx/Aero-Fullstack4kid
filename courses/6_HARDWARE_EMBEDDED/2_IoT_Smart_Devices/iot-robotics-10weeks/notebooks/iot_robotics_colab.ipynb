### ⚡ Tuần 1 & 2: Mô Phỏng Đọc Tín Hiệu Analog ADC & Bộ Lọc Nhiễu

```python
import numpy as np
import matplotlib.pyplot as plt

# Sinh tín hiệu điện áp Analog 12-bit ADC (0 - 4095) bị nhiễu
np.random.seed(42)
raw_adc = np.random.normal(2048, 150, 100)

# Bộ lọc trung bình động (Moving Average Filter)
window_size = 5
filtered_adc = np.convolve(raw_adc, np.ones(window_size)/window_size, mode='valid')

plt.figure(figsize=(10, 4))
plt.plot(raw_adc, label='Raw ADC Input (Noisy)', color='lightgray')
plt.plot(filtered_adc, label='Moving Average Filtered Output', color='blue', linewidth=2)
plt.xlabel('Sample')
plt.ylabel('ADC Value (0-4095)')
plt.title('ESP32 12-bit ADC Signal Filtering Simulation')
plt.legend()
plt.grid(True)
plt.show()
```

### 🌐 Tuần 6: Mô Phỏng MQTT Telemetry Publisher qua Public Broker

```python
import paho.mqtt.client as mqtt
import json, time

BROKER = "broker.hivemq.com"
PORT = 1883
TOPIC = "aero_fullstack4kid/iot/telemetry"

client = mqtt.Client("Colab_IoT_Publisher")
client.connect(BROKER, PORT, 60)

# Gửi 3 gói tin telemetry JSON
for i in range(3):
    payload = {
        "device_id": "ESP32_WOKWI_01",
        "temperature": round(25.0 + i * 1.2, 2),
        "humidity": round(60.0 + i * 0.5, 2)
    }
    client.publish(TOPIC, json.dumps(payload))
    print(f"[+] Published to {TOPIC}: {payload}")
    time.sleep(1)

client.disconnect()
```

### 🤖 Tuần 8: Mô Phỏng Động Học Xe Robot 2 Bánh (Differential Drive)

```python
def diff_drive_kinematics(v_L, v_R, L=0.15):
    """Tính vận tốc dài v và vận tốc góc omega từ tốc độ 2 bánh."""
    v = (v_R + v_L) / 2.0
    omega = (v_R - v_L) / L
    return v, omega

# Thử nghiệm: Bánh phải 0.5 m/s, Bánh trái 0.2 m/s
v, w = diff_drive_kinematics(0.2, 0.5)
print(f"[AMR KINEMATICS] Linear Velocity v : {v:.2f} m/s")
print(f"[AMR KINEMATICS] Angular Velocity w: {w:.2f} rad/s")
```
