### ⚡ Tuần 2: Mô Phỏng ADC 12-bit & Đèn LED PWM Dimmer

```python
import numpy as np
import matplotlib.pyplot as plt

# Mô phỏng giá trị đọc ADC 16-bit MicroPython (0 - 65535)
adc_raw_values = np.linspace(0, 65535, 100)
voltages = (adc_raw_values / 65535.0) * 3.3 # Điện áp 3.3V
pwm_duties = adc_raw_values # Thang duty PWM

plt.figure(figsize=(8, 4))
plt.plot(adc_raw_values, voltages, label='Calculated Voltage (V)', color='green')
plt.xlabel('MicroPython read_u16() Raw Value')
plt.ylabel('Voltage (V)')
plt.title('Raspberry Pi Pico RP2040 ADC Scaling Characteristic')
plt.grid(True)
plt.show()
```
