### 📷 Tuần 4 & 5: Mô Phỏng Thuật Toán Dò Vạch Làn Đường Hough Lines

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Tạo ảnh sa hình tổng hợp (Synthetic Road Image)
img = np.zeros((480, 640, 3), dtype=np.uint8)
# Vẽ vạch làn trái (White) và làn phải (Yellow)
cv2.line(img, (120, 480), (260, 240), (255, 255, 255), 10)
cv2.line(img, (520, 480), (380, 240), (0, 255, 255), 10)

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
blurred = cv2.GaussianBlur(gray, (5, 5), 0)
edges = cv2.Canny(blurred, 50, 150)

# Hough Lines Transform
lines = cv2.HoughLinesP(edges, 1, np.pi/180, 20, minLineLength=20, maxLineGap=50)

output = img.copy()
if lines is not None:
    for l in lines:
        x1, y1, x2, y2 = l[0]
        cv2.line(output, (x1, y1), (x2, y2), (0, 255, 0), 3)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Synthetic Road Image')
plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(output, cv2.COLOR_BGR2RGB))
plt.title('Hough Lines Detected Lanes')
plt.show()
```

### 🤖 Tuần 8: Mô Phỏng Mạng Nơ-ron CNN TFLite Bám Làn Đường

```python
import tensorflow as tf
from tensorflow.keras import layers, models

# Xây dựng mô hình CNN bám làn đường siêu nhẹ
model = models.Sequential([
    layers.Conv2D(24, (5, 5), strides=(2, 2), activation='relu', input_shape=(66, 200, 3)),
    layers.Conv2D(36, (5, 5), strides=(2, 2), activation='relu'),
    layers.Flatten(),
    layers.Dense(50, activation='relu'),
    layers.Dense(1)
])

model.compile(optimizer='adam', loss='mse')
print("[+] Keras Self-Driving CNN Model Summary:")
model.summary()
```
