### 🚁 Tuần 1: Tính Toán Lực Nâng Quadcopter & Tỷ Lệ Thrust-to-Weight

```python
def calculate_drone_thrust(total_weight_g, num_motors=4):
    """Tính toán lực nâng tối thiểu cần thiết cho từng động cơ (Thrust-to-Weight Ratio = 2:1)"""
    required_total_thrust = total_weight_g * 2.0 # 2:1 ratio for stable hovering
    thrust_per_motor = required_total_thrust / num_motors
    return required_total_thrust, thrust_per_motor

weight = 850 # 850 grams
total_req, per_motor = calculate_drone_thrust(weight)
print(f"[DRONE PHYSICS] Total Vehicle Weight: {weight} g")
print(f"[DRONE PHYSICS] Required Total Thrust (2:1 Ratio): {total_req:.1f} g")
print(f"[DRONE PHYSICS] Minimum Required Thrust Per Motor : {per_motor:.1f} g")
```

### 📡 Tuần 6: Giả Lập Tin Nhắn MAVLink Telemetry

```python
import time
import numpy as np
import matplotlib.pyplot as plt

# Mô phỏng dữ liệu Roll / Pitch góc nghiêng Drone thu được từ Flight Controller MAVLink
timestamps = np.linspace(0, 10, 100)
roll_angles = 5.0 * np.sin(2 * np.pi * 0.5 * timestamps) + np.random.normal(0, 0.5, 100)
pitch_angles = 3.0 * np.cos(2 * np.pi * 0.5 * timestamps) + np.random.normal(0, 0.5, 100)

plt.figure(figsize=(10, 4))
plt.plot(timestamps, roll_angles, label='Roll Angle (deg)', color='red')
plt.plot(timestamps, pitch_angles, label='Pitch Angle (deg)', color='blue')
plt.xlabel('Time (s)')
plt.ylabel('Angle (degrees)')
plt.title('Simulated MAVLink Flight Telemetry (Attitude Data)')
plt.legend()
plt.grid(True)
plt.show()
```
