### 🤖 CEH Mod 20: AI SOC Threat Detection

```python
import numpy as np, pandas as pd
from sklearn.ensemble import IsolationForest

data = pd.DataFrame({'reqs': np.concatenate([np.random.normal(50, 5, 95), np.random.normal(500, 50, 5)])})
clf = IsolationForest(contamination=0.05).fit(data[['reqs']])
data['anomaly'] = clf.predict(data[['reqs']])
print(data[data['anomaly'] == -1])
```
